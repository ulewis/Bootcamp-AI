```python
import numpy as np

# Definir la función de activación escalón
def step_function(x):
    return 1 if x >= 0 else 0

# Definir la función de activación tanh
def tanh_activation(x):
    return np.tanh(x)

# Definir la función del perceptrón
def perceptron(inputs, weights, bias, activation_func):
    # Calcular la suma ponderada de las entradas
    weighted_sum = np.dot(inputs, weights) + bias
    # Aplicar la función de activación
    output = activation_func(weighted_sum)
    return output


```

## ¿Qué factores influyen en una enfermedad cardiaca?

```python

# Datos de prueba: presión arterial y colesterol
presion_arterial = 100  # Ejemplo de presión arterial sistólica, Probar con diastólica
colesterol = 50       # Ejemplo de nivel de colesterol de lipoproteínas de alta densidad

# Pesos y sesgo (bias) del perceptrón
weights = np.array([0.5, -0.5])  # Pesos arbitrarios para las características
bias = -30                  # Sesgo arbitrario

# Establecemos las entradas del perceptrón
inputs = np.array([presion_arterial, colesterol])
```

## Hacemos funcionar al perceptrón

```python

# Aplicar el perceptrón con función de activación escalón
output_step = perceptron(inputs, weights, bias, step_function)

# Aplicar el perceptrón con función de activación tanh
output_tanh = perceptron(inputs, weights, bias, tanh_activation)
```

## En los resultados podemos observar la diferencia de utilizar diferentes funciones de activación

```python
# Imprimir resultados
print("Resultado usando función de activación escalón:")
print("Valor de salida:", output_step)
print("El paciente tiene enfermedad cardíaca." if output_step == 1 else "El paciente no tiene enfermedad cardíaca.")

print("\nResultado usando función de activación tanh:")
print("Valor de salida:", output_tanh)
print("El paciente tiene enfermedad cardíaca." if output_tanh >= 0 else "El paciente no tiene enfermedad cardíaca.")
```

## Perceptrón Tipo AND, OR Y XOR

### Intenemos hacer un perceptrón tipo AND

```python
# Función para probar diferentes combinaciones de pesos y funciones de activación
def test_perceptron(inputs, weights, bias, activation_func):
    print("Prueba con pesos:", weights)
    print("Función de activación:", activation_func.__name__)
    for p, q in [(0, 0), (0, 1), (1, 0), (1, 1)]:
        input_data = np.array([p, q])
        prediction = perceptron(input_data, weights, bias, activation_func)
        print(f'Entradas: ({p}, {q}), Predicción: {prediction}')

# Datos de entrada para las pruebas
inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

# Probando diferentes combinaciones de pesos y funciones de activación
weights_1 = np.array([0.4, 0.4])
bias_1 = -0.5
test_perceptron(inputs, weights_1, bias_1, step_function)

```

```python

weights_2 = np.array([0.8, 0.5])
bias_2 = -0.7
test_perceptron(inputs, weights_2, bias_2, step_function)
```

### Ahora un tipo OR

```python
#Intenta hacer un OR jugando con los pesos y bias
weights_3 = np.array([2, 1])
bias_3 = -0.5
test_perceptron(inputs, weights_3, bias_3, step_function)
```

```python
import numpy as np
import tensorflow as tf

x = np.array([-2.0, -1.0, 0.0, 1.0, 2.0])  # Array de ejemplo

# Calcular la tangente hiperbólica usando NumPy
tanh_np = np.tanh(x)
print("Tanh con NumPy:", tanh_np)

# Convertir el array NumPy a un tensor TensorFlow
x_tf = tf.constant(x)

# Calcular la tangente hiperbólica usando TensorFlow
tanh_tf = tf.math.tanh(x_tf)
print("Tanh con TensorFlow:", tanh_tf.numpy())
```
